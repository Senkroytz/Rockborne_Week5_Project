```python
import random
import time

# Below variables are constant 
MIN_NUM = 1
MAX_NUM = 10
ALLOWED_ATTEMPTS = 3

# Below are the global variables 
wins = 0
losses = 0
successful_tries = 0
total_tries = 0


def gen_num():
    # generates a random number in range MIN_NUM and MAX_NUM inclusive
    generated_number = random.randint(MIN_NUM, MAX_NUM)
    return generated_number


def timing(start_time, end_time):
    # stopwatch timer used for timing duration of each attempt
    time_taken = end_time - start_time
    return time_taken


def win_rate():
    # calculates player win rate
    global wins, losses

    if wins + losses == 0:
        return 0
    win_percent = (wins/(wins + losses))
    return win_percent


def accuracy():
    # calculates player accuracy score
    global successful_tries, total_tries

    if total_tries == 0:
        return 0
    percent_accuracy = (successful_tries / total_tries)
    return percent_accuracy


def play_game():
    # contains the game logic
    global wins, losses, successful_tries, total_tries

    while True:
        rand_num = gen_num()
        attempts = ALLOWED_ATTEMPTS
        tries = 0
        start_time = time.time()

        while attempts > 0:
            try:
                guess = int(input(f"\nGuess a number between {MIN_NUM} and {MAX_NUM} inclusive ({attempts} attempt(s) left): "))
            except ValueError:
                # Check if input is integer
                print("Please enter a valid integer.")
                continue

            if guess < MIN_NUM or guess > MAX_NUM:
                # checks if a guess is in range of MIN_NUM and MAX_NUM
                print(f"This number is beyond the range of {MIN_NUM} and {MAX_NUM}")
                continue

            tries += 1
            total_tries += 1
            attempts -= 1

            if guess == rand_num:
                # logic for when a guess is correct
                wins += 1
                successful_tries += 1

                end_time = time.time()

                print(f"\nCorrect! The number was {rand_num}.")
                print(f"You guessed it in {tries} attempt(s) and took {timing(start_time, end_time):.2f} second(s).")
                print(f"Your win rate is {win_rate():.2%} based on {wins + losses} game(s)")
                print(f"Your accuracy rate is {accuracy():.2%} based on {total_tries} tries")
                break

            elif guess > rand_num:
                print("Incorrect, try a smaller number.")
            elif attempts > 0:
                print("Incorrect, try a bigger number.")

        else:
            # logic for when 0 attempts are remaining
            losses += 1
            print(f"\nIncorrect, you ran out of guesses. The number was {rand_num}.")
            print(f"Your win rate is {win_rate():.2%} based on {wins + losses} game(s)")
            print(f"Your accuracy rate is {accuracy():.2%} based on {total_tries} tries")

        while True:
            # input field for whether a play wants to continue playing or not
            play_again = input("Play again (y/n)? ").lower()
            if play_again in ["y", "n"]:
                break
            print("Please type 'y' to play again or 'n' to stop.")

        if play_again == "n":
            break


if __name__ == "__main__":
    # main function
    global wins, losses, successful_tries, total_tries

    print("Welcome to my number guessing game!")
    print(f"You will be given {ALLOWED_ATTEMPTS} attempts to guess a number between {MIN_NUM} and {MAX_NUM} inclusive")
    print("You will be scored based on your round win rate and guess accuracy")
    input("Good luck! Press Enter to start ")
    play_game()
    print(f"Thank you for playing! Final result: {wins} win(s) and {losses} loss(es).")
    print(f"Final accuracy: {accuracy():.2%}")
```
